# ระบุจำนวน record ที่หายไป

```python
import calendar

missing_data_monthly = []

# Group by station, year, and month
for (station_name, year, month), group in rainall_filtered.groupby(['tele_station_oldcode', 'year', 'month']):
    total_days_in_month = calendar.monthrange(year, month)[1]
    measured_days_count = group['measure_datetime'].nunique()
    missing_days_count = total_days_in_month - measured_days_count
    missing_days_proportion = missing_days_count / total_days_in_month

    missing_data_monthly.append({
        'Station_Name': station_name,
        'Year': year,
        'Month': month,
        'Missing_Days_Count': missing_days_count,
        'Missing_Days_Prop': missing_days_proportion
    })

daylost_monthly = pd.DataFrame(missing_data_monthly)
display(daylost_monthly.head(10))
display(daylost_monthly.tail(10))
display(daylost_monthly.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter data for Month 8
month_8_data = daylost_monthly[daylost_monthly['Month'] == 8]

# Sort the data by Missing_Days_Prop for better visualization
month_8_data_sorted = month_8_data.sort_values(by='Missing_Days_Prop', ascending=False)

# Select only the top 50 stations
month_8_data_top50 = month_8_data_sorted.head(50)

# Create the bar chart for the top 50 stations
plt.figure(figsize=(15, 8))
sns.barplot(x='Station_Name', y='Missing_Days_Prop', data=month_8_data_top50)
plt.xticks(rotation=90)
plt.title('Top 50 Stations with Highest Missing Days Proportion in Month 8')
plt.xlabel('Station Name')
plt.ylabel('Missing Days Proportion')
plt.tight_layout()
plt.show(50)
```

```python
daliyrain_use = daliyrain_filtered[['station_code','measure_datetime','data']]
daliyrain_use
```

```python
daliyrain_use['measure_datetime'] = pd.to_datetime(daliyrain_use['measure_datetime'])
daliyrain_use['day'] = daliyrain_use['measure_datetime'].dt.day
daliyrain_use['year'] = daliyrain_use['measure_datetime'].dt.year
daliyrain_use['month'] = daliyrain_use['measure_datetime'].dt.month

display(daliyrain_use.head())
```

ตัดข้อมูลเอาเฉพาะในเดือน 7 8 9 10 11 12 มา เพื่อลดขนาด และคิด Median ของวันนั้นๆในแต่ละปี

```python
daliyrain_predict= daliyrain_use[(daliyrain_use['month'].isin([7, 8, 9, 10, 11, 12]))]
display(daliyrain_predict.head())
display(daliyrain_predict.shape)
```

# เติมค่า Median

**ทำไมต้องเติม Median**

เนื่องจากในบางปีอาจจะเกิดฝนตกหนักมากในวันนั้นๆทั้งๆที่ปกติเทียบจะไม่ตกเลย ซึ่งมีปัจจัยหลายสาเหตุ เช่น เกิดเอลนีโญ่ที่ทำให้ฝนถล่ม ทำให้ค่าในปีนั้นสูงเกินปกติจนเกิดOutliner เมื่อนำมาคิดค่าเฉลี่ยอาจทำให้ค่าสูงเกินความเป็นจริง

จึงเลือกใช้ Median ที่มีความเสถียรมากกว่า สะท้อนความปกติวันในนั้นๆของปี

```python
# Calculate the median rainfall for each day of each month across all years
daily_monthly_median_rain = daliyrain_predict.groupby(['month', 'day'])['data'].median().reset_index()
daily_monthly_median_rain.rename(columns={'data': 'median_rainfall'}, inplace=True)

display(daily_monthly_median_rain.head())
display(daily_monthly_median_rain.tail())
display(daily_monthly_median_rain.shape)
```

```python
# 1. Extract 'day' from 'measure_datetime' in rainall_filtered_final
rainall_filtered_final['day'] = rainall_filtered_final['measure_datetime'].dt.day

# 2. Merge rainall_filtered_final with daily_monthly_median_rain to get median_rainfall for each day/month
rainall_filled = pd.merge(
    rainall_filtered_final,
    daily_monthly_median_rain,
    on=['month', 'day'],
    how='left'
)

# 3. Fill NaN values in 'rainfall_value' with 'median_rainfall'
# Ensure that only NaN values are filled. If median_rainfall is also NaN, it will remain NaN.
rainall_filled['rainfall_value'] = rainall_filled['rainfall_value'].fillna(rainall_filled['median_rainfall'])

# 4. Drop the temporary 'median_rainfall' and 'day' columns if no longer needed
rainall_filtered_final = rainall_filled.drop(columns=['median_rainfall', 'day'])

# Display info and check for nulls in rainfall_value again
print("Rainfall values after filling NaNs:")
display(rainall_filtered_final.head())
display(rainall_filtered_final.isnull().sum())
```

# สรุปผล


```python
monthly_rainfall_summary = rainall_filtered_final.groupby(['tele_station_oldcode','month','year'])['rainfall_value'].sum().reset_index()
monthly_rainfall_summary.rename(columns={'tele_station_oldcode': 'Node', 'rainfall_value': 'rain'},
                                inplace=True)

rainfall_summary = monthly_rainfall_summary[['Node', 'rain', 'month', 'year']]
rainfall_summary
```
